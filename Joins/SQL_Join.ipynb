
# 📘 What is Sql Join ?

- A JOIN is used to combine rows from two or more tables based on a related column between them.

- Joins are mainly used in SELECT queries to fetch data from multiple tables.

- Commonly, the foreign key in one table matches the primary key in another table.

- Used for fetch data from two or more tables



<br>

## SQL Keywords for Joins (Combining Tables)

| Keyword      | Description                                                |
| ------------ | ---------------------------------------------------------- |
| `JOIN`       | Combines rows from two or more tables                      |
| `INNER JOIN` | Returns records with matching values in both tables        |
| `LEFT JOIN`  | Returns all records from left table + matched from right   |
| `RIGHT JOIN` | Returns all records from right table + matched from left   |
| `FULL JOIN`  | Returns all records where there is a match in either table |
| `ON`         | Specifies join condition                                   |
| `UNION`      | Combines results of two SELECT queries (no duplicates)     |
| `UNION ALL`  | Combines results including duplicates                      |

## 🟡 2. LEFT JOIN / LEFT OUTER JOIN
- When we combine two or more table using left join, it returns:
- All rows from the left table

- Matching rows from the right table

- For non-matching rows → right table columns become NULL

#### Syntaxt:


    SELECT columns
    FROM table1 t1
    LEFT JOIN table2 t2
    ON t1.col = t2.col;

## 🟣 3. RIGHT JOIN / RIGHT OUTER JOIN
- It returns all rows from the right table

- Only matching rows from the left table

- Non-matching left rows → become NULL

#### Syntax:

    SELECT columns
    FROM table1 t1
    RIGHT JOIN table2 t2
    ON t1.col = t2.col;

## 🔴 4. FULL JOIN / FULL OUTER JOIN
- It return all the matching rows from both tables

- Non-matching rows from both sides
- Missing values are filled with NULL

#### Syntax:

    SELECT columns
    FROM table1 t1
    FULL JOIN table2 t2
    ON t1.col = t2.col;

## 🟢 5. CROSS JOIN

- It returns Cartesian Product → every row of table1 is paired with every row of table2.

-  If table1 has 3 rows and table2 has 4 rows → output = 3 × 4 = 12 rows

- ⭐ `No ON condition allowed.`

#### Syntax:

    SELECT *
    FROM table1
    CROSS JOIN table2;

## ⚫ 6. SELF JOIN (Extra Important for Interviews)

- It is when a table is joined with itself.
- it id s regular join where a table is joined with itself
- this technique is used to compare or relate rows within
- the same table, treating it as two temporary separate
- tables by assigning different alias names to each
- instance.

#### Syntax:

    SELECT e1.name AS Employee, e2.name AS Manager
    FROM employee e1
    INNER JOIN employee e2
    ON e1.manager_id = e2.emp_id;

# Joins Summary Table

<br>

| JOIN Type  | Matching Rows    | Unmatched Rows                     | NULL Filling?      |
| ---------- | ---------------- | ---------------------------------- | ------------------ |
| INNER JOIN | Only matching    | No                                 | No                 |
| LEFT JOIN  | All from left    | Non-matching from left → included  | right table = NULL |
| RIGHT JOIN | All from right   | Non-matching from right → included | left table = NULL  |
| FULL JOIN  | All rows         | All                                | both sides = NULL  |
| CROSS JOIN | All combinations | N/A                                | No                 |

#🔷 UNION

- Removes duplicate rows

- Both SELECT statements must have:

- Same number of columns

- Same order

- Similar datatypes

#### Syntax:

    SELECT col FROM table1
    UNION
    SELECT col FROM table2;

# 🔥 Differnce between Union vs Join




| Feature                    | JOIN                                                  | UNION                                              |
| -------------------------- | ----------------------------------------------------- | -------------------------------------------------- |
| **What it combines**       | Combines tables **side-by-side**                      | Combines results **top-to-bottom**                 |
| **Output shape**           | More **columns**                                      | More **rows**                                      |
| **Requires relationship?** | ✔ Yes (except CROSS JOIN)                             | ❌ No                                               |
| **Condition needed?**      | Requires **ON** condition (except NATURAL/CROSS JOIN) | No condition required                              |
| **Used for**               | Showing related data from multiple tables             | Merging results of multiple SELECT queries         |
| **Column requirement**     | Columns can be different                              | Number & datatype of columns must be **same**      |
| **Duplicate handling**     | Returns all matches (duplicates possible)             | `UNION` removes duplicates, `UNION ALL` keeps them |
| **Example meaning**        | Show Customer + their Orders                          | Show all names from Customer + Supplier together   |
| **Performance**            | Generally heavier join operations                     | Faster, simple row append                          |
| Work                | Horizontally  | Vertically
