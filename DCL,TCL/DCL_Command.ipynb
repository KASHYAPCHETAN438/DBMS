# 🧑‍💻 User in Mysql

## 🔹 Mysql Database
- The user table inside the mysql database is a system table that stores all MySQL user accounts and their privileges.

🧩 Key Points:

- It contains login details (username, host, password) and permissions for every MySQL user.

- It decides who can connect, from where, and what actions they can perform.

- It is part of the internal system database mysql, created automatically when MySQL is installe



<br>


### Syntax :    


     -- View all details (User name, Host Type ,previleges)
     USE mysql;
     show tables;
     SELECT * FROM user;

                        OR

     -- View only User name and Host Type
     SELECT User, Host FROM mysql.user;

## 1. Create a local user


### Syntax :
     
     CREATE USER 'username'@'host' IDENTIFIED BY 'password';

<br>

### **Explanation of Each Part:**

| Part                           | Meaning                                                                                     |
| ------------------------------ | ------------------------------------------------------------------------------------------- |
| **`CREATE USER`**              | This command tells MySQL that you are creating a new user.                                  |
| **`'username'`**               | The name of the new MySQL user you want to create (e.g., `'mohit'`).                        |
| **`'host'`**                   | Specifies **from where the user can connect** (like `'localhost'`, `'%'`, `'192.168.1.5'`). |
| **`IDENTIFIED BY 'password'`** | Sets the **login password** for the user (it’s stored in encrypted form).                   |

## 👉When ever we create user By default, MySQL automatically gives Previliges

- GRANT USAGE ON `*.*` TO `mohit`@`localhost`

### This means:

- Mohit can connect to MySQL (after entering password).

- But Mohit cannot SELECT, INSERT, UPDATE, or DELETE any data.

- Basically, a valid user account but with no data access

# 📘 What is DCL ?

- DCL stands for Data Control Language
- DCL commands in SQL are used to control access or permissions to the database objects like tables, views, and schemas.
- They help in managing user privileges and security within the database.

## SQL Keywords for Data Control (DCL)

| Command    | Description                                                | Syntax                                                 | Example                                      |
| ---------- | ---------------------------------------------------------- | ------------------------------------------------------ | -------------------------------------------- |
| **GRANT**  | Gives specific privileges (permissions) to users or roles. | `GRANT privilege_name ON object_name TO user_name;`    | `GRANT SELECT, INSERT ON Students TO Mohit;` |
| **REVOKE** | Removes (takes back) privileges given to users.            | `REVOKE privilege_name ON object_name FROM user_name;` | `REVOKE INSERT ON Students FROM Mohit;`      |

<br>



<br>

## 🔐 Privileges (Permissions)

| Privilege          | Description                                  |
| ------------------ | -------------------------------------------- |
| **SELECT**         | User can retrieve (read) data from a table.  |
| **INSERT**         | User can insert new records into a table.    |
| **UPDATE**         | User can modify existing data.               |
| **DELETE**         | User can remove data from a table.           |
| **ALL PRIVILEGES** | Grants all possible permissions.             |
| **REFERENCES**     | User can create foreign key references.      |
| **ALTER**          | User can change table structure.             |
| **INDEX**          | User can create or remove indexes.           |
| **EXECUTE**        | User can run stored procedures or functions. |

# ------------- GRANT COMMAND -------------

## 1. GRANT Command (To Give single Permission)

### Syntax :    

    GRANT privilege_name
    ON object_name
    TO user_name;

            OR

    GRANT privilege_name
    ON database_name.table_name
    TO user_name;

### Example :    

    GRANT SELECT ON sql_practices.Students TO 'mohit'@'localhost';
