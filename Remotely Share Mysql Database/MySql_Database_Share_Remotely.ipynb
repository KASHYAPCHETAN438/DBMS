# STEP - 4 &nbsp;&nbsp; CHANGE BIND ADDRESS


## **What is bind-address?**

- bind-address is a MySQL server setting inside my.ini / mysqld.cnf.

- It tells MySQL which IP address it should listen on for incoming connections.

- It controls whether MySQL accepts only local connections or remote connections.

- By default bind address is 127.0.0.1 which is address of localhost means you can connect mysql server only by local host not remotely

<br>

###✅ **Why Change bind-address to 0.0.0.0?**
1. Allows Remote Access

- 127.0.0.1 = MySQL listens only on your own PC

  → Remote system cannot connect ❌

- 0.0.0.0 = MySQL listens on all network interfaces

  → Any system in the network can connect ✔ (if user has access)

<br>

### **How to change Bind Address ?**

- Find the MySQL configuration file (my.ini)


     
      XAMPP MySQL location
      C:\xampp\mysql\bin\my.ini

               OR

      MySQL Installer location
      C:\ProgramData\MySQL\MySQL Server 8.0\my.ini
      
- Open it in notepad and serach Bind address

- Then change `Bind Address = 0.0.0.0`
