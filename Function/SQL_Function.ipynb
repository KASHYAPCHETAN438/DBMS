# 📘 SQL Function

- In SQL, functions are predefined (built-in) operations that perform tasks on data — such as calculations, string handling, date/time processing, or data conversion.

- It take input (arguments) and return a single value.

<br>

## 🧠 Types of SQL Functions

#### SQL functions are mainly divided into two categories:

<br>

### 1. **Scalar Functions**

- Work on one row at a time

- Return one value per row

- Examples: UPPER(), LOWER(), ROUND(), LENGTH(), NOW()

<br>

### 2. **Group (Aggregate) Functions**

- Work on a group of rows

- Return one value for the group

- Examples: COUNT(), SUM(), AVG(), MIN(), MAX()

<br><br>

<br>

# 🧩 1. Scalar Function Examples

## (a) String Functions

| Function                        | Description             | Example                                               |
| ------------------------------- | ----------------------- | ----------------------------------------------------- |
| `UPPER(str)`                    | Converts to uppercase   | `SELECT UPPER('mohit');` → **MOHIT**                  |
| `LOWER(str)`                    | Converts to lowercase   | `SELECT LOWER('MOHIT');` → **mohit**                  |
| `LENGTH(str)`                   | Returns string length   | `SELECT LENGTH('Mohit');` → **5**                     |
| `CONCAT(a,b)`                   | Joins two strings       | `SELECT CONCAT('Hello ', 'Mohit');` → **Hello Mohit** |
| `SUBSTRING(str, start, length)` | Extracts part of string | `SELECT SUBSTRING('MohitKashyap', 1, 5);` → **Mohit** |

## (b) Numeric Functions

| Function        | Description                   | Example                                |
| --------------- | ----------------------------- | -------------------------------------- |
| `ROUND(num, n)` | Rounds a number to n decimals | `SELECT ROUND(25.678, 2);` → **25.68** |
| `CEIL(num)`     | Rounds up                     | `SELECT CEIL(4.2);` → **5**            |
| `FLOOR(num)`    | Rounds down                   | `SELECT FLOOR(4.8);` → **4**           |
| `POWER(a,b)`    | a raised to b                 | `SELECT POWER(2,3);` → **8**           |

## (c) Date Functions

| Function          | Description         | Example                                         |
| ----------------- | ------------------- | ----------------------------------------------- |

## (D) Control Flow Functions

| Function / Keyword    | Description                            | Example                                        |
| --------------------- | -------------------------------------- | ---------------------------------------------- |
| `IF(condition, a, b)` | Returns *a* if condition true else *b* | `SELECT IF(salary>50000,'High','Low');`        |
| `IFNULL(a, b)`        | Returns *a* if NOT NULL else *b*       | `SELECT IFNULL(bonus,0);`                      |
| `NULLIF(a, b)`        | Returns NULL if both values equal      | `SELECT NULLIF(10,10);` → **NULL**             |
| `CASE`                | Multiple conditions (like switch-case) | `CASE WHEN marks>=90 THEN 'A' ELSE 'Fail' END` |

# 🧩 2. What is SQL Aggregate Functions ?

#### - Aggregate functions perform a calculation on a set of values and return a single value.
#### - They are often used with the GROUP BY clause to summarize data.

<br>

## **Aggregate (Group) Functions**

#### - Aggregate functions are special SQL functions that perform a calculation on a set of rows and return a single summarized value.

<br>


| Function  | Description   | Example                          |
| --------- | ------------- | -------------------------------- |
| `COUNT()` | Counts rows   | `SELECT COUNT(*) FROM Students;` |
| `SUM()`   | Adds values   | `SELECT SUM(Age) FROM Students;` |
| `AVG()`   | Average value | `SELECT AVG(Age) FROM Students;` |
| `MIN()`   | Minimum value | `SELECT MIN(Age) FROM Students;` |
| `MAX()`   | Maximum value | `SELECT MAX(Age) FROM Students;` |
